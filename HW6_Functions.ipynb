**1. Функція з параметром**

Створіть функцію `show_menu()`, яка приймає перелік страв, наприклад ['Піца', 'Паста', 'Салат'] і виводить меню ресторану в наступному вигляді:
```
=== МЕНЮ ===
1. Піца
2. Паста
3. Салат
=============
```

Викличте цю функцію на списку ["Суп грибний", "Котлета по-київськи", "Лимонад"].

````python
# Спосіб 1
def show_menu(menu_items):
  print("=== МЕНЮ ===")
  for index, item in enumerate(menu_items, start=1):
    print(f"{index}. {item}")
  print("=============")

cafe_menu = ["Суп грибний", "Котлета по-київськи", "Лимонад"]
show_menu(cafe_menu)
````

````python
# Спосіб 2
def show_menu(menu_items):
  menu_to_show = "=== МЕНЮ ===\n"
  for index, item in enumerate(menu_items, start=1):
    menu_to_show = menu_to_show + f"{index}. {item}\n"
  menu_to_show = menu_to_show + "============="
  print(menu_to_show)

cafe_menu = ["Суп грибний", "Котлета по-київськи", "Лимонад"]
show_menu(cafe_menu)
````

**2. Функція з параметром**

Створіть функцію `print_square(n)`, яка виводить квадрат числа.

Наприклад: `print_square(5)` → "Квадрат числа 5 = 25"

Викличте її з різними числами.

````python
def print_square(n):
  square = n ** 2
  print(f"Квадрат числа {n} = {square}")

print_square(2)
print_square(5)
print_square(10)
print_square(11)
print_square(3)
````

**4. Функція пошуку максимумального значення**

Створіть функцію `find_max(numbers)`, яка:
- Приймає список чисел
- Повертає найбільше число в списку

Протестуйте на списку `[15, 42, 8, 23, 56, 19]`.


````python
def find_max(numbers):
  return max(numbers)

my_list = [15, 42, 8, 23, 56, 19]
max_value = find_max(my_list)
print(max_value)
````

**5. Функція з умовами**

Створіть функцію `check_grade(grade)`, яка:
- Приймає оцінку (число)
- Повертає "Здав" якщо grade >= 60
- Повертає "Не здав" якщо grade < 60

Протестуйте на різних оцінках.


````python
import json
````

````python
def check_grade(grade):
  if grade >= 60:
    return "Здав"
  else:
    return "Не здав"

my_grade =  73
my_result = check_grade(my_grade)
print(my_result)

student_grades = {
    "Математика": 62,
    "Англійська мова": 80,
    "Фізика": 59,
    "Біологія": 60,
    "Хімія": 57,
    "Географія": 72
}

student_results = {}

for subject, grade in student_grades.items():
  student_results[subject] = check_grade(grade)


print(student_results)

print(json.dumps(student_results, indent=4, ensure_ascii=False))

def show_my_progress(results):
  print("==== MY PROGRESS ====")
  for subject, result in results.items():
    print(f"{subject}: {result}")
  print("=====================")

show_my_progress(student_results)
````
